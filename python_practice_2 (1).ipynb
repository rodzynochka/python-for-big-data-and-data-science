19. Створіть дві множини. Продемонстуйте над ними операції: обʼєднання, різниці, пересічі та симетричної різниці. Використовуйте методи, що не змінюють множини, а створюють нові.

```python
set_1 = {"music", "reading", "singing", "hiking"}
set_2 = {"cycling", "hiking", "music", "dancing", "climbing"}
set_union = set_1.union(set_2)
set_diff = set_1.difference(set_2)
set_inter = set_1.intersection(set_2)
set_symmetr = set_1.symmetric_difference(set_2)
print("Union:", set_union)
print("Difference:", set_diff)
print("Intersection:", set_inter)
print("Symmetric difference:", set_symmetr)
```
