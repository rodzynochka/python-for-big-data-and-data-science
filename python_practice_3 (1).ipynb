3. (13б) Створіть рядок зі своїм імʼям латинкою. Напишіть код, який би автоматизовував наступну задачу.

Якщо в імені є літера "а"
 або "i", Виведіть на екран повідомлення "A or I is in the name".



Якщо в імені є літери "o" та "n", Виведіть на екран повідомлення "O and N is in the name".

Також якщо в імені немає літери "о", але є "n", Виведіть на екран повідомлення "Only N is in the name".

В решті випадків виведіть на екран повідомлення "There is no N in the name, but there might be O".

Зверніть увагу, що, наприклад, для імені "Antonina" буде виведено два повідомлення.

```python
my_name = "Anna"

if "a" in my_name.lower() or "i" in my_name.lower():
    print("A or I is in the name")
elif "o" in my_name.lower() and "n" in my_name.lower():
    print("O and N is in the name")
elif "n" in my_name.lower() and not "o" in my_name.lower():
    print("Only N is in the name")
else:
    print("There is no N in the name, but there might be O")
```

5. (9б) Створіть словник, використовуючи dict comprehension, що містить if else. Коли слід використовувати comprehensions, а коли їх слід уникати?

```python
new_dict = {n: n**2 if n%2 == 0 else n+0.5
            for n in range(0,13)}
print(new_dict)
```

*місце для відповіді: Comprehensions варто використовувати, коли потрібно створити список, словник або множину, застосовуючи прості умови або фільтрацію. Однак, якщо логіка занадто складна, містить багато умов або змінює зовнішні змінні, краще використати звичайні цикли для кращої зрозумілості.

9. (13б) Створіть список з елементами булевого типу або None. Використовуючи цикли, отримайте в результаті список з кортежами, де перший елемент кортежу - індекс(ціле число), а другий елемент - відповідне значення з першого списку.

Наприклад, [True, True, None, False] -> [(0, True), (1, True), (2, None), (3, False)]

```python
new_list = [False, True, True, None]

result = [(index, value) for index, value in enumerate(new_list)]

print(result)
```

12. (13б) Гра "Нумо вгадай".
Розробіть просту гру, де користувач має вгадати випадкове число від 1 до 100. Для генерації випадкового числа використовуйте функцію random.randint(a, b) (тут a та b включно). Для отримання числа з консолі використовуйте функцію input(), результат якої обовʼязково явно приведіть до типу int.

Підказка: використовуйте цикл while та умови if-else, щоб повідомити користувачу піказки (напр. "Більше", "Менше") поки користувач не вгадає. По завершенню виведіть кількість спроб, які знадобилися для вгадування.

```python
import random
attempts = 0

while True:
  random_num = random.randint(1, 100)
  num = int(input("Type a number from 1 to 100: "))
  attempts += 1

  if num < 1:
    print("Can't be less than 1")
    continue
  elif num > 100:
    print("Can't be more than 100")
    continue
  elif num == random_num:
    print(f"You guessed the number, the number was {random_num}, it took you {attempts} attempts!")
    break

  print(f"You did not guess the number, the number was {random_num}")
```
