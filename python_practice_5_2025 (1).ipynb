6. (11б). Створіть 2 дочірніх класи від 3d-фігура: Циліндр та Куб, які мають атрибути, необхідні для розрахунку об'єму. Визначте метод get_volume в кожному з них так, щоби вони розраховували об'єм у залежності від формули для кожного типу фігури. Створіть обʼєкт класу Циліндр та обʼєкт класу Куб. Виконайте операцію суми за допомогою оператора + між ними, аби знайти сумарний об'єм.

````python
import math

class Figure3D:
    def get_volume(self):
        return 0

    def __add__(self, other):
        volume = self.get_volume() + other.get_volume()

class Cylinder(Figure3D):
    def __init__(self, radius, height):
        self.radius = radius
        self.height = height

    def get_volume(self):
        return math.pi * self.radius ** 2 * self.height

class Cube(Figure3D):
    def __init__(self, side):
        self.side = side

    def get_volume(self):
        return self.side ** 3

cylinder = Cylinder(4, 6)
cube = Cube(5)

print(f"Cylinder volume: {cylinder.get_volume():.2f}")
print(f"Cube volume: {cube.get_volume():.2f}")

total_volume = cylinder.get_volume() + cube.get_volume()
print(f"Total volume of the figures is: {total_volume:.2f}")
````

8. (13б). Створіть клас EWallet з приватними атрибутами balance та id. Реалізуйте методи поповнення та зняття коштів, забезпечивши належну інкапсуляцію. Підказка: використовуйте декоратори property та setter.

````python

class EWallet:
    def __init__(self, wallet_id, initial_balance=0):
        self.__id = wallet_id
        self.__balance = initial_balance

    @property
    def balance(self):
        return self.__balance

    @property
    def id(self):
        return self.__id

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            return True
        return False

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            return True
        return False

    def __str__(self):
        return f"EWallet(id={self.__id}, balance={self.__balance})"

wallet = EWallet("wallet Anna1234", 100000)
print(wallet.balance)
print(wallet.id)
print(wallet.deposit(100))
print(wallet.balance)
print(wallet.withdraw(150))
print(wallet.balance)
````

9. (11б). Створіть клас ShoppingCart, який містить список об'єктів типу Item. Реалізуйте методи для додавання айтемів, їх видалення та відображення списку айтемів.

````python
class Item:

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return f"{self.name} – ${self.price:.2f}"


class ShoppingCart:

    def __init__(self):
        self.items = []

    def __add__(self, item):
        self.items.append(item)
        print(f"{item.name} was added to the shopping cart")

    def remove_item(self, item_name):
        for item in self.items:
            if item.name == item_name:
                self.items.remove(item)
                print(f"{item.name} was deleted from the shopping cart")
                return
        print(f"{item_name} was not found in the shopping cart")

    def show_items(self):
        if not self.items:
            print("The cart is empty")
        else:
            print("Items in the shopping cart:")
            for item in self.items:
                print(f"- {item}")

my_cart = ShoppingCart()
item_1 = Item("Milk", 40)
item_2 = Item("Banana", 35)

my_cart.__add__(item_1)
my_cart.__add__(item_2)
my_cart.show_items()
my_cart.remove_item("Apple")
my_cart.remove_item("Banana")
my_cart.show_items()
````

Ітератори — це об'єкти, що реалізують методи __iter__() та __next__() і дозволяють по черзі отримувати елементи ітерабельного об'єкта.

13. (11б). Створіть клас, що має визначений Iterator Protocol та при кожному виклику next повертає назву дня тижня, поки вони не вичерпаються. Зверніть увагу, що порядок має значення.

````python
class Weekdays:

    def __init__(self):
        self.days = ("Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday")
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
       if self.index >= len(self.days):
          print("oops, the week has ended :)")
       else:
          day = self.days[self.index]
          self.index += 1
          return day

week_iter = Weekdays()

print(next(week_iter))
print(next(week_iter))
print(next(week_iter))
print(next(week_iter))
print(next(week_iter))
print(next(week_iter))
print(next(week_iter))
print(next(week_iter))

````
